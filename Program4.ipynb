```python
import pandas as pd
df = pd.read_csv("/content/train_and_test2.csv")
print(df)
```

```python
data = {
'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena', 'Arjun'],
'Age': [21, 22, None, 23, 21, None],
'Salary': [25000, None, 30000, 28000, None, 32000],
'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}
df = pd.DataFrame(data)
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
print(df)
```

```python
data = {
'Student': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena'],
'Department': ['CSE', 'ECE', None, 'ISE', None]
}
df = pd.DataFrame(data)
print("Before Forward Fill:")
print(df)
df['Department'] = df['Department'].ffill()
print("\nAfter Forward Fill:")
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
data = {
'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}
df = pd.DataFrame(data)
encoder = LabelEncoder()
df['Department_Encoded'] = encoder.fit_transform(df['Department'])
print(df)
print(encoder.classes_)
```

```python
import pandas as pd
data = {
'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}
df = pd.DataFrame(data)
encoded_df = pd.get_dummies(df, columns=['Department'])
print(encoded_df)
```
